```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy 
import PIL
import requests
```

```python
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0, 100, 50)

fig, ax = plt.subplots(figsize=(5,3)) #fig is the whole thing, the figure. 5 and 3 are the size, 5 by 3 inches.
ax.scatter(x,y) #ax is the axes, just the subplot area with the data points.
```

```python
fig, ax = plt.subplots(figsize=(5,3)) #fig is the whole thing, the figure. 5 and 3 are the size, 5 by 3 inches.
ax.scatter(x,y) #Having both scatter and bar with overlay both.
ax.bar(x,y) 
```

```python
#If you want to change the plot, you have to respecify it.

fig, ax = plt.subplots(figsize=(5,3)) #fig is the whole thing, the figure. 5 and 3 are the size, 5 by 3 inches.
ax.bar(x,y) #Doing a bar graph instead of scatter
```

```python
#If you want to change the plot, you have to respecify it.

fig, ax = plt.subplots(figsize=(5,3)) #fig is the whole thing, the figure. 5 and 3 are the size, 5 by 3 inches.
ax.plot(x,y) #Doing a line graph.
```

```python
#If you want to change the plot, you have to respecify it.

fig, ax = plt.subplots(figsize=(5,3)) #fig is the whole thing, the figure. 5 and 3 are the size, 5 by 3 inches.
ax.hist(y)
```

```python
#Adding labels.

fig, ax = plt.subplots(figsize=(5,3)) #fig is the whole thing, the figure. 5 and 3 are the size, 5 by 3 inches.
ax.plot(x,y) #Doing a line graph.
ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')
fig.tight_layout() #This will change the dimensions and properties to make it tighter and rmove white space
```

```python
#Can modify fonts and what not. Here are making font dictionaries.

font1 = {'family':'sans-serif','color':'blue','size':20}
font2 = {'family':'monospace','color':'green','size':14}
```

```python
#Changing font properties. Need to pre-specify your font dictionaries.

fig, ax = plt.subplots(figsize=(5,3)) #fig is the whole thing, the figure. 5 and 3 are the size, 5 by 3 inches.
ax.plot(x,y) #Doing a line graph.
ax.set_title('Total growth over time', fontdict=font1, loc='left') #Loc is location, it's setting the alignment here to be left-aligned. Can also do centre, and right.
ax.set_ylabel('Total growth', fontdict=font2)
ax.set_xlabel('Years since start', fontdict=font2)
fig.tight_layout() #This will change the dimensions and properties to make it tighter and rmove white space
```

```python
fig, ax = plt.subplots(figsize=(5,3)) #fig is the whole thing, the figure. 5 and 3 are the size, 5 by 3 inches.
ax.scatter(x,y, marker ='*', color ='indigo') #Changing properties of the dots.
#fig.show() #This is print. this is the default in python, but may need to tell it to print in the python type files (.py). Since this is jupytr you don't need this line (showing is the default).
```

```python
fig, ax = plt.subplots(figsize=(5,3)) #fig is the whole thing, the figure. 5 and 3 are the size, 5 by 3 inches.
ax.plot(x,y,marker ='*', color ='indigo', linestyle = '--', linewidth = 2) #Changing properties of the lines. Can also add dots to a line plot
```

```python
fig, ax = plt.subplots(figsize=(5,3)) 
ax.plot(x,y,marker ='*', color ='#7425b9', linestyle = '--', linewidth = 2) #Can also specify colour via hexcode.
```

```python
fig, ax = plt.subplots(figsize=(5,3)) 
ax.plot(x,y,marker ='*', color ='#7425b9', linestyle = '--', linewidth = 2,markeredgecolor= '#fa9359', markerfacecolor = '#000000') 
#Can also specify the properties of the marker outlines, etc. Just like shape, shape outline, shape fill in excel.
```

```python
fig, ax = plt.subplots(figsize=(5,3)) 
ax.plot(x,y,marker ='*', markersize = 12, color ='#7425b9', linestyle = '--', linewidth = 2,markeredgecolor= '#fa9359', markerfacecolor = '#000000')
ax.grid(axis = 'y', color = 'blue', linewidth = 2, linestyle = '--') 
#Can also add grid lines. axis is horizontal vs vertical lines. Can specify properties just like the other stuff.
```

```python
fig,ax=plt.subplots(figsize=(5,3))
ax.grid(axis='y',color='blue',linewidth=2,linestyle='-.')
ax.plot(x,y,marker='*',markersize=12,color='#7425b9',linestyle='--',linewidth=2,markeredgecolor='#fa9359',markerfacecolor='#000000')
```
