```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy 
import PIL
import requests
```

```python
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0, 100, 50)
y2 = np.random.randint(0, 100, 50)
```

```python
fig, ax = plt.subplots(figsize=(5,3)) 
ax.plot(x,y1,y2) #Doing a line graph with 2 variables.
```

```python
#Code from class, makes the same plot.
fig, ax = plt.subplots(figsize=(5,3)) 
ax.plot(x,y1)
ax.plot(x,y2)
```

```python
#Adding legend.
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc='lower right',
frameon = True)
```

```python
#Adding legend.
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc='lower right',
frameon = True, shadow = False)
```

```python
#Adding legend.
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc='lower right',
frameon = False, shadow = True, fontsize = 12, ncol = 2) #Shadow only shows if frame exists.
```

```python
#Moving the legend outside of the figure.
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc='upper left',
bbox_to_anchor =(1, 1)) #Attaching the legends upper left location to the plots top right corner so they are aligned/attached.

#loc is important to specify which aspect of the plot is the reference. So if you put centre, then it would be attaching the center to the coordinates instead of the top left.
#If you just want the default inside the graph, you can put just loc.
#0,0 may be bottom left corner. 1, 1 is the top right corner. Can use incremental (0.5, 0.8) to adjust the location more specifically.
```

```python
#Specifying text
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.text(10, 95, 'This value is important!', ha = 'center', color = 'red', size = '20') #10, 95 is the coordinate, 0,0 is bottom left.
```

```python
#Specifying text
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.text(25, 50, 'This value is important!', ha = 'center', color = 'green', size = '10') #I think this coordinate is based on the left-most point of the text.
```

```python
#Make blank axes and add text in reference to different components.
fig,ax=plt.subplots()
ax.axis([0,10,0,10])
ax.text(1,5,". Data:(1,5)",transform=ax.transData)
ax.text(0.5, 0.1, ". Axes:(0.5, 0.1)",   transform=ax.transAxes)    
ax.text(0.2, 0.2, ". Figure:(0.2, 0.2)", transform=fig.transFigure)
```

```python
#Can also annotate with arrows.
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.annotate('This is important!', xy=(10, 95), xytext=(20, 94),
arrowprops=dict(facecolor='black'))
#Dictionary is required to spcify more elements.
```

```python
#Can also annotate with arrows.
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.annotate('This is important!', xy=(10, 95), xytext=(30, 70), #xy specifies where the point of the arrow is, and text is seperate (changing those relatively changes the shape of the arrow.)
arrowprops=dict(facecolor='black'))
#Dictionary is required to specify more elements like arrow style.
#See slides for arrow styles. https://matplotlib.org/3.1.1/api/_as_gen/matplotlib.pyplot.annotate.html
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.annotate('This is important!', xy=(10,95),xytext=(20,94),arrowprops=dict(arrowstyle = "wedge", color = "green"))
```

```python
#Removing tick marks and labels
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.yaxis.set_major_locator(plt.NullLocator()) #Removes labels AND ticks on the y axis.
ax.xaxis.set_major_formatter(plt.NullFormatter()) #Removes labels but not ticks.
```

```python
#Changing bin size.
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.xaxis.set_major_locator(plt.MaxNLocator(10)) #Specify how many evenly spaced bins you want.
```

```python
#Changing axis interval size.
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.xaxis.set_major_locator(plt.MultipleLocator(10)) #Specify bins of a particular size.
```

```python
#Changing axis interval size.
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
plt.xticks(rotation=45,ha='right') #Rotating the x-axis lables 45degrees, alighning the right of the text. It will always target the current axes (active subplot).
```
