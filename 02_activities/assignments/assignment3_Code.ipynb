```python
#Import packages.
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
#Load file.
df = pd.read_csv("C:/Users/Elia/Dropbox/Data Sciences Certificate/visualization/02_activities/assignments/mtcg-all-population-groups-library-activities-en-2023.csv")
print(df)
```

```python
#Because data transformation is out of the scope of this assignment, I am selecting only libraries A to B (the first 42 rows).
df_graph = df.loc[0:41]
df_graph['Annual Music Downloads / Borrows'] = df_graph['Annual Music Downloads / Borrows'].astype(int)
df_graph
```

```python
#Making the graph.
sns.set_style('whitegrid')
sns.set_theme(rc = {'figure.figsize':(6,8)})
graph = sns.scatterplot(data=df_graph, x='Annual Music Downloads / Borrows', y ='Library Name', hue = "Annual Music Downloads / Borrows")
graph.set(title='Annual Music Use by Library', xlabel='Annual Music Use', ylabel='Library')
plt.legend(loc='upper left', bbox_to_anchor=(1, 1), title='# of Borrows/Downloads')
```
